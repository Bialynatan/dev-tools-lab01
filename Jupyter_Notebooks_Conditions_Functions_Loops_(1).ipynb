## Functions

### How to Define a Function

When creating a function, it is mandatory to follow a certain order in syntax. The function syntax is as follows:

* A function definition always starts with the keyword "def", followed by the function name and parentheses.
* All input parameters or arguments should be placed within those parentheses. In parentheses, one can also define the parameters. Parameters are optional.
* First, one can place the documentation string (optional).
* The code block of each function starts with a colon (:) and is indented.
* The return statement exits the function, and one can optionally pass back an expression to the caller. A return statement with no arguments is equivalent to return None.

```python
# Example of a function in Python

def my_function(name):
    print("Hi " + name)

my_function("Ankur")
```

```python
# Function returning values

def add_function(a, b):
    return a+b

a = 8
b = 12
c = add_function(a, b)

print(c)
```

```python
# Function variable scope

x = 98 # Global variable scope

def f1():
    print(x)

def f2():
    print(x)

f1()
f2()
```

```python
# Function variable scope

x = 98 # Global variable scope

def f1():
    x = 15 # local variable
    print(x)

def f2():
    x = 30 # local variable
    print(x)

f1()
f2()
print(x)
```

```python
# Function variable scope

x = 98 # Global variable scope

def f1():
    global x
    x = 50
    print(x)

def f2():
    x = 30 # local variable
    print(x)

f1()
f2()
print(x)
```

```python
# Creating a function without default parameters

def intro(name,city,interest):
    intro = 'Hi I am {},I live in {} and I love {}'.format(name,city,interest)
    return intro

intro("Alice","Mumbai","Music")

intro("Ryan","Delhi") # Missing argument will return an error
```

```python
# Creating a function with default parameters
def intro(name,city,interest="Music"):
    intro = 'Hi I am {},I live in {} and I love {}'.format(name,city,interest)
    return intro

intro("Ryan","Delhi")

```

```python
intro("Ryan","Delhi", "Football")
```
